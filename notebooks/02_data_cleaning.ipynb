```python
import pandas as pd

customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")

order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")

products = pd.read_csv("../data/raw/olist_products_dataset.csv")

sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")

order_payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")

order_reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")

geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")

category_translation = pd.read_csv(
    "../data/raw/product_category_name_translation.csv"
)
```

```python
customers_clean = customers.copy()
orders_clean = orders.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
order_reviews_clean = order_reviews.copy()
order_items_clean = order_items.copy()
order_payments_clean = order_payments.copy()
geolocation_clean = geolocation.copy()
category_translation_clean = category_translation.copy()
```

```python
order_reviews_clean[
    order_reviews_clean['review_id'].duplicated(keep=False)
].sort_values('review_id').head(20)
```

```python
order_reviews_clean['review_comment_title'] = (
    order_reviews_clean['review_comment_title'].fillna('')
)

order_reviews_clean['review_comment_message'] = (
    order_reviews_clean['review_comment_message'].fillna('')
)
```

```python
products_clean[
    products_clean['product_category_name'].isna()
][[
    'product_id',
    'product_category_name',
    'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]].head(10)
```

```python
products_clean[
    products_clean['product_weight_g'] == 0
][[
    'product_id',
    'product_category_name',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]]
```

```python
order_payments_clean[
    order_payments_clean['payment_value'] <= 0
][[
    'order_id',
    'payment_type',
    'payment_installments',
    'payment_value'
]]
```

```python
orders_clean.groupby('order_status')[
    [
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date'
    ]
].apply(lambda x: x.isna().sum())
```

```python
order_reviews_clean['review_comment_title'] = (
    order_reviews_clean['review_comment_title'].astype(str).str.strip()
)

order_reviews_clean['review_comment_message'] = (
    order_reviews_clean['review_comment_message'].astype(str).str.strip()
)
```

```python
products_clean['product_category_name'] = (
    products_clean['product_category_name'].str.strip()
)
```

```python
customers_clean['customer_city'] = (
    customers_clean['customer_city'].str.strip()
)

customers_clean['customer_state'] = (
    customers_clean['customer_state'].str.strip()
)
```

```python
sellers_clean['seller_city'] = (
    sellers_clean['seller_city'].str.strip()
)

sellers_clean['seller_state'] = (
    sellers_clean['seller_state'].str.strip()
)
```

```python
geolocation_clean['geolocation_city'] = (
    geolocation_clean['geolocation_city'].str.strip()
)

geolocation_clean['geolocation_state'] = (
    geolocation_clean['geolocation_state'].str.strip()
)
```

```python
category_translation_clean['product_category_name'] = (
    category_translation_clean['product_category_name'].str.strip()
)

category_translation_clean['product_category_name_english'] = (
    category_translation_clean['product_category_name_english'].str.strip()
)
```

```python
orders_clean.to_csv("../data/cleaned/orders_clean.csv", index=False)
customers_clean.to_csv("../data/cleaned/customers_clean.csv", index=False)
products_clean.to_csv("../data/cleaned/products_clean.csv", index=False)
sellers_clean.to_csv("../data/cleaned/sellers_clean.csv", index=False)
order_reviews_clean.to_csv("../data/cleaned/order_reviews_clean.csv", index=False)
order_items_clean.to_csv("../data/cleaned/order_items_clean.csv", index=False)
order_payments_clean.to_csv("../data/cleaned/order_payments_clean.csv", index=False)
geolocation_clean.to_csv("../data/cleaned/geolocation_clean.csv", index=False)
category_translation_clean.to_csv("../data/cleaned/category_translation_clean.csv", index=False)
```

```python
print("orders:", len(orders), len(orders_clean))
print("customers:", len(customers), len(customers_clean))
print("products:", len(products), len(products_clean))
print("sellers:", len(sellers), len(sellers_clean))
print("order_reviews:", len(order_reviews), len(order_reviews_clean))
print("order_items:", len(order_items), len(order_items_clean))
print("order_payments:", len(order_payments), len(order_payments_clean))
print("geolocation:", len(geolocation), len(geolocation_clean))
print("category_translation:", len(category_translation), len(category_translation_clean))
```
